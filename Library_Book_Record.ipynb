```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages
from google.colab import files
import re

# Step 1: Upload Excel file
uploaded = files.upload()
```

```python
# Step 2: Load Excel Sheet
file_name = "Library Book Record-08-04-2025.xlsx"
df = pd.read_excel(file_name, sheet_name="Exam-LIB-11-4-2025")
```

```python
df = df[df['BOOK TYPE'].astype(str).str.strip().str.upper() != 'TEACHER COPY']
df
```

```python
valid_classes = ['NUR', 'K1', 'K2'] + [str(i) for i in range(1, 13)]

def to_additional_if_invalid(value):
    if pd.isna(value):
        return 'ADDITIONAL BOOKS'
    val = str(value).strip().upper()

    # Reject ranges like "1 TO 2", "5 & 6", etc.
    if re.search(r'\d+\s*(TO|&|-)\s*\d+', val):
        return 'ADDITIONAL BOOKS'

    # Allow only single valid entries like "5", "K2", etc.
    return val if val in valid_classes else 'ADDITIONAL BOOKS'

df['CLASS'] = df['CLASS'].apply(to_additional_if_invalid)
df
```

```python
df['SUBJECT'] = df['SUBJECT'].astype(str).str.strip().str.upper()
df['BOOK SUB TYPE'] = df['BOOK SUB TYPE'].astype(str).str.strip().str.upper()
df['SUBJECT_COMBO'] = df['SUBJECT'] + ' - ' + df['BOOK SUB TYPE']
df
```

```python
grouped = df.groupby(['SUBJECT_COMBO', 'CLASS'], dropna=False).size().reset_index(name='No. of Books')
pivot_df = grouped.pivot(index='SUBJECT_COMBO', columns='CLASS', values='No. of Books').fillna(0).astype(int).reset_index()
pivot_df
```

```python
# Step 5: Save to Excel
output_file = "libraryDataNew3.xlsx"
pivot_df.to_excel(output_file, index=False)

# Step 6: Download the file
files.download(output_file)
```
